```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_excel(r"C:\Users\Alen Jose\Downloads\myexcel.xlsx")

# Preprocessing: Replace "Height" column with random values
df['Height'] = np.random.randint(150, 181, size=len(df))

# Task 1: Distribution of employees across teams
team_distribution = df['Team'].value_counts()
team_percentage = (team_distribution / len(df) * 100).round(2)

# Visualization for Task 1
plt.figure(figsize=(10, 6))
team_distribution.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Distribution of Employees Across Teams', fontsize=14)
plt.xlabel('Team', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()

# Task 2: Segregate employees based on positions
position_distribution = df['Position'].value_counts()

# Visualization for Task 2
plt.figure(figsize=(10, 6))
position_distribution.plot(kind='bar', color='coral', edgecolor='black')
plt.title('Distribution of Employees by Position', fontsize=14)
plt.xlabel('Position', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()

# Task 3: Identify the predominant age group
bins = [18, 25, 35, 45, 60]
labels = ['18-25', '26-35', '36-45', '46-60']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
age_group_distribution = df['Age Group'].value_counts()

# Visualization for Task 3
plt.figure(figsize=(8, 6))
age_group_distribution.plot(kind='bar', color='limegreen', edgecolor='black')
plt.title('Predominant Age Group of Employees', fontsize=14)
plt.xlabel('Age Group', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.tight_layout()
plt.show()

# Task 4: Team and position with highest salary expenditure
team_salary = df.groupby('Team')['Salary'].sum()
position_salary = df.groupby('Position')['Salary'].sum()
highest_team_salary = team_salary.idxmax()
highest_position_salary = position_salary.idxmax()

# Visualization for Task 4
plt.figure(figsize=(10, 6))
team_salary.sort_values().plot(kind='bar', color='teal', edgecolor='black')
plt.title('Salary Expenditure by Team', fontsize=14)
plt.xlabel('Team', fontsize=12)
plt.ylabel('Total Salary ($)', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
position_salary.sort_values().plot(kind='bar', color='purple', edgecolor='black')
plt.title('Salary Expenditure by Position', fontsize=14)
plt.xlabel('Position', fontsize=12)
plt.ylabel('Total Salary ($)', fontsize=12)
plt.xticks(rotation=45, fontsize=10)
plt.tight_layout()
plt.show()

# Task 5: Correlation between age and salary
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Age', y='Salary', color='orange', edgecolor='black')
plt.title('Correlation Between Age and Salary', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Salary ($)', fontsize=12)
plt.tight_layout()
plt.show()

# Display insights
print("Insights:")
print(f"1. Team with the highest salary expenditure: {highest_team_salary}")
print(f"2. Position with the highest salary expenditure: {highest_position_salary}")
print(f"3. Percentage distribution of employees across teams:\n{team_percentage}")
print(f"4. Predominant age group:\n{age_group_distribution.idxmax()}")
```
